```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer

```

```python
df = pd.read_csv('Datasets/covid_toy.csv')
df.head()
```

```python
df.isnull().sum()   #there are missing values in fever will impute them :)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['has_covid']), df['has_covid'], test_size=0.2, random_state=2)
```

```python
X_train.head() #will apply ordinal on 'cough' amd onehot on 'gender'and 'city' and impute on 'fever'
```

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('t1', SimpleImputer(),['fever']),
    ('t2', OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('t3', OneHotEncoder(sparse_output=False, drop='first'),['gender','city'])
],remainder='passthrough')# remainder will keep the other columns as it is or drop them if we set remainder='drop'
```

```python
transformer.fit_transform(X_train).shape
```
