```python
import matplotlib.pyplot as plt
%matplotlib inline

import numpy as np

x = np.arange(0,10)
y = np.arange(11,21)

plt.scatter(x, y, c='r')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Scatter Plot Example')
plt.savefig('scatter_plot.png')# Save the figure as a PNG file
```

```python
y = x*x
plt.plot(x, y, 'b--')
```

```python
#subplot example
# 2 rows, 2 columns, plot number 1(index)
plt.subplot(2,2,1)
plt.plot(x, y, 'r--')
plt.subplot(2,2,2)
plt.plot(y, x, 'g*-')
plt.subplot(2,2,3)
plt.plot(x, y, 'bs-')
plt.subplot(2,2,4)
plt.plot(y, x, 'r^-')


```

```python
x=np.arange(1,11)
y = 2*x +5
plt.plot(x, y, 'g-')
```

```python
#sin wave
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
plt.plot(x, y, c='b')
```

```python
#bar plot
x = [2,4,6,8,10]
y = [1,3,5,7,9]
x2 = [1,3,5,7,9]
y2 = [2,4,6,8,10]
plt.bar(x2, y2, color='b')
plt.bar(x, y, color='g')
plt.xlabel('X axis')
plt.ylabel('Y axis')    
plt.title('Bar Plot Example')
```

```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a, bins=[0,20,40,60,80,100])
plt.title('Histogram Example')
plt.xlabel('Value') 
plt.ylabel('Frequency')
```

```python
#Box plot


'''
 Components of a Boxplot
- Minimum: The smallest data point (excluding outliers).
- First Quartile (Q1): The 25th percentile — lower edge of the box.
- Median (Q2): The 50th percentile — line inside the box.
- Third Quartile (Q3): The 75th percentile — upper edge of the box.
- Maximum: The largest data point (excluding outliers).
- Whiskers: Lines extending from the box to the min and max values.
- Outliers: Data points outside 1.5×IQR (interquartile range) from Q1 or Q3, often shown as dots.
'''

group1 = np.random.normal(60, 10, 100)
group2 = np.random.normal(50, 15, 100)

plt.boxplot([group1, group2], labels=["Group 1", "Group 2"])
plt.title("Comparison of Two Groups")
plt.show()
```

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]
plt.boxplot(data, vert=True, patch_artist=True)
```

```python
#pie chart
labels = 'A', 'B', 'C', 'D'
sizes = [15, 30, 45, 10]
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0, 0, 0.1, 0)  # explode 1st slice

plt.pie(sizes, explode = explode, labels = labels, colors = colors, autopct = '%1.1f%%')
```

```python
import seaborn as sns

df = sns.load_dataset('tips')
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
df.dtypes
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

```python
#jointplot
#used for univariate and bivariate analysis
sns.jointplot(x='total_bill', y='tip', data=df, kind='hex')
```

```python
sns.jointplot(x='total_bill', y='tip', data=df, kind='reg')
#in this will see a regression line and  density line at histogram with plot
```

```python
sns.pairplot(df)
#there are more than one features in dataset so it will plot pairplot for all features
```

```python
df['sex'].value_counts()
```

```python
sns.pairplot(df,hue='sex')
```

```python
#distplot
sns.histplot(df['tip'], kde=True)
```

```python
sns.histplot(df['tip'], kde=False,bins=10)
```

```python
#categorical plot
sns.countplot(x='sex',data=df)
```

```python
sns.countplot(x='smoker',data=df)
```

```python
sns.countplot(x='day',data=df)
```

```python
#Bar plot
sns.barplot(x='day', y='total_bill', data=df)
```

```python
#Box plot or wisker plot
#It is used to show distribution of data based on five number summary( minimum, first quartile, median, third quartile, maximum)
#and there are outliers also shown in boxplot
sns.boxplot(x='smoker', y='total_bill', data=df)
```

```python
#box plot of all the features in dataset
sns.boxplot(data=df,orient='v')
```

```python
#categorizing data based on some other categories   
sns.boxplot(x='day', y='total_bill', hue='smoker', data=df)
```

```python
#violin plot
#help us to see both the distribution of data in terms of kernel density estimation and the box plot representation
sns.violinplot(x='total_bill', y='day', data=df)
```
