```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('Datasets/train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
X.head()
```

```python
#applying cross val score to the data
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

```python
#Constructing new feature

X['Family_size'] = X['SibSp'] + X ['Parch'] + 1
X.head()
```

```python
def func(num):
    if num == 1:
        #alone
        return 0
    elif num > 1 and num <= 4:
        #small family
        return 1
    else :
        #large family
        return 2
    
```

```python
X['Family_type'] = X['Family_size'].apply(func)
```

```python
X.head()
```

```python
#now we dont requre the columns beside Family type
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
#accuracy gets a slight increase
```
